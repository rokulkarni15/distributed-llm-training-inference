# Inference Generator

```python


# Cell 1: Install dependencies if needed (run once)
# !pip install torch transformers peft accelerate bitsandbytes  # For quantization if VRAM-limited

# Cell 2: Interactive model loading
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer
from peft import PeftModel
import os

# Base model name
base_model_name = "meta-llama/Llama-2-7b-hf"

# Fine-tuned path - adjust to your actual path
peft_model_path = "./checkpoints/zero2_1gpu/final"  # e.g., zero2_1gpu or zero2_4gpu

# Interactive choice
choice = input("Load model: 'base' for raw Hugging Face model or 'finetuned' for local fine-tuned? ").strip().lower()
if choice not in ['base', 'finetuned']:
    raise ValueError("Please enter 'base' or 'finetuned'.")

# Load tokenizer (from base or fine-tuned; same for both)
if choice == 'finetuned' and os.path.exists(peft_model_path):
    tokenizer = AutoTokenizer.from_pretrained(peft_model_path)
else:
    tokenizer = AutoTokenizer.from_pretrained(base_model_name)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

# Load model
device = "cuda" if torch.cuda.is_available() else "cpu"
model = None

if choice == 'finetuned':
    if not os.path.exists(peft_model_path):
        raise FileNotFoundError(f"Fine-tuned path not found: {peft_model_path}")
    # Load base
    base_model = AutoModelForCausalLM.from_pretrained(
        base_model_name,
        torch_dtype=torch.float16 if device == "cuda" else torch.float32,
        device_map="auto" if device == "cuda" else None,
        trust_remote_code=True
    )
    # Load PEFT LoRA
    model = PeftModel.from_pretrained(base_model, peft_model_path)
    loaded_from = f"Fine-tuned from: {peft_model_path}"
else:
    # Load raw base model
    model = AutoModelForCausalLM.from_pretrained(
        base_model_name,
        torch_dtype=torch.float16 if device == "cuda" else torch.float32,
        device_map="auto" if device == "cuda" else None,
        trust_remote_code=True
    )
    loaded_from = f"Raw base model: {base_model_name}"

model = model.to(device)
model.eval()  # Inference mode

print(f"Model loaded successfully! {loaded_from}")
```

```python
# Cell 3: Interactive inference function (unchanged from before)
def generate_response(prompt, max_new_tokens=256, temperature=0.7, top_p=0.9):
    """
    Generate a response from the loaded model.
    
    Args:
    - prompt: Input text (e.g., "Write a Python function to sort a list by key.")
    - max_new_tokens: Max tokens to generate
    - temperature: Sampling creativity (0.1=deterministic, 1.0=creative)
    - top_p: Nucleus sampling threshold
    """
    # Tokenize input
    inputs = tokenizer(prompt, return_tensors="pt", truncation=True, max_length=512)
    inputs = {k: v.to(device) for k, v in inputs.items()}
    
    # Generate
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            temperature=temperature,
            top_p=top_p,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id,
            eos_token_id=tokenizer.eos_token_id,
            repetition_penalty=1.1  # Avoid loops
        )
    
    # Decode response (full output minus input)
    full_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
    response = full_text[len(prompt):].strip()  # Extract generated part
    
    return response

# Example usage in Jupyter
user_prompt = input("Enter your prompt: ")  # e.g., "Explain how to fine-tune Llama with LoRA."
response = generate_response(user_prompt)
print("\nGenerated Response:")
print(response)

# For batch: responses = [generate_response(p) for p in ["Prompt1", "Prompt2"]]
```

```python
# Cell: Unload model and free memory
import gc
import torch

# Delete model and tokenizer objects
if 'model' in globals():
    del model
if 'base_model' in globals():
    del base_model
if 'tokenizer' in globals():
    del tokenizer

# Force garbage collection
gc.collect()

# Clear CUDA cache if using GPU
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()
    
print("Model unloaded and memory freed!")

# Optional: Check memory status
if torch.cuda.is_available():
    print(f"GPU memory allocated: {torch.cuda.memory_allocated() / 1024**3:.2f} GB")
    print(f"GPU memory reserved: {torch.cuda.memory_reserved() / 1024**3:.2f} GB")
```
